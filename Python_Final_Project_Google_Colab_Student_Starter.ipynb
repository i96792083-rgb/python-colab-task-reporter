## Step 1 — Prepare the sample data
This cell creates a JSON file inside the current Colab runtime. **Run this first** whenever you restart the runtime. Files generated in the Colab runtime may disappear after the session ends.

```python
import json
import re
import requests

sample_tasks = [
    {"id": 1, "title": "Review Python functions", "completed": True},
    {"id": 2, "title": "Practise list comprehensions", "completed": False},
    {"id": 3, "title": "Read an API response", "completed": True},
    {"id": 4, "title": "Create a Python class", "completed": True},
    {"id": 5, "title": "Publish the project on GitHub", "completed": False},
]

with open("sample_tasks.json", "w", encoding="utf-8") as file:
    json.dump(sample_tasks, file, indent=2)

print("Created sample_tasks.json")
```

## Step 2 — Read the JSON file
**Exercise 1:** Load `sample_tasks.json` using `json.load()` and print how many records it contains.
**Expected:** `Loaded 5 tasks`

```python
# TODO: Read the JSON file into the variable records.
# Hint: use with open(..., "r", encoding="utf-8")

# print(f"Loaded {len(records)} tasks")

with open ("sample_tasks.json", "r", encoding = "utf-8") as file:
  records = json.load(file)

print(f"Loaded {len(records)} tasks")
```

## Step 3 — Define a class
**Exercise 2:** Create a class named `Task`. Each object needs a `title` and a `completed` value. Add a method `status()` that returns `"Done"` or `"Pending"`. Then create a list of `Task` objects from the data you loaded.

**Hint:** Use `for row in records:` to keep this step easy.

```python
class Task:
    def __init__(self, title, completed):
        self.title = title
        self.completed = completed
    def status(self):
        return "Done" if self.completed else "Pending"
tasks = []
for row in records:
    tasks.append(Task(row["title"], row["completed"]))
    print(tasks[0].title, "-" , tasks[0].status())
```

## Step 4 — Validate a course code
**Exercise 3:** Use a regular expression to check that the course code has the format `PY-101`: the letters `PY-`, then exactly three digits. For an invalid code, raise `ValueError` and handle it with `try` / `except`.

**Test:** change `course_code` to `PY101` to see the error message, then restore `PY-101`.

```python
course_code = "PY-101"

def check_course_code(code):
    return re.fullmatch(r"PY-\d{3}", code) is not None
try:
    if not check_course_code(course_code):
        raise ValueError("Invalid course code")
    print("Course code:", course_code)
except ValueError as error:
    print("Error:", error)
```

## Step 5 — Build a summary with a list comprehension
**Exercise 4:** Create a list of the **titles** for tasks that are completed. Use one list comprehension with `if`. Then build a report dictionary with course code, totals and titles.

```python
completed_titles = [task.title for task in tasks if task.completed]
report = {
    "course_code": course_code,
    "total": len(tasks),
    "completed": len(completed_titles),
    "pending": len(tasks) - len(completed_titles),
    "completed_titles": completed_titles,
}
print("Course:", report["course_code"])
print("Total tasks:", report["total"])
print("Completed:", report["completed"])
print("Pending:", report["pending"])
print("Completed task titles:")
for title in completed_titles:
    print("-", title)
```

## Step 6 — Write the report to a JSON file
**Exercise 5:** Save the `report` dictionary as `progress_report.json` using `json.dump(..., indent=2)`. Reopen the file with `json.load()` to confirm the values.

```python
with open("progress_report.json", "w", encoding="utf-8") as file:
    json.dump(report, file, indent=2)
with open("progress_report.json", "r", encoding="utf-8") as file:
    saved_report = json.load(file)
print("Saved: ", saved_report["course_code"])
```

## Step 7 — Try one public API request
